```python
def percent_of_total(s):
    return np.round(s / sum(s) * 100, 2)
```

```python
percent_of_total(np.arange(3,15,3))
```

```python
def hypotenuse(x, y):
    return (x**2 + y**2) ** .5
```

```python
hypotenuse(9,12)
```

### Function to return age given a year

```python
def age(year):
    age = 2026 - year
    return age
```

```python
age(1942)
```

```python
def name_and_age(name, year):
    '''
    Returns a string with name and age of person.
    Arguements:
            name: name of the person
            year: year the person was born
    returns:
        string with name and age
    '''
    return name + ' is ' + str(age(year)) + ' years old.'
```

```python
name_and_age('noah', 2004)
```

```python
ages = Table().with_columns(
    'person', make_array('jim', 'pam', 'joe', 'mary'),
    'birth year', make_array(1985, 1988, 1967, 1904)
)
ages
```

```python
ages.apply(name_and_age, 'person', 'birth year')
```

```python
ages.apply(age, 'birth year')
```
