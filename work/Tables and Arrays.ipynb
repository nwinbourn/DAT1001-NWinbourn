# Ways to create a table

## creating a table from scratch

```python
#creating columns
streets = make_array('Harmony', 'Horsetooth', 'drake', 'prospect')
streets
```

```python
south = Table().with_columns('street name', streets)
south
```

```python
type(south)
```

```python
type(streets)
```

```python
south=south.with_column('blocks from campus', np.arange(0,4))
south
```

```python
south = south.with_row(['Mulburry', 5])
south
```

```python
south = south.with_column('blocks from campus', np.where(south.column('street name') == 'Harmony', 4, south.column('blocks from campus'))
)
south
```

```python
south=south.with_column('city', 'Fort Collins')
```

```python
south
```

## Accessing a cell in a table

```python
south
```

```python
south.column('blocks from campus').item(4)
```

```python
south.row(4).item('blocks from campus')
```

```python
south.where('street name', 'Mulburry').take(0)['blocks from campus'].item(0)
```

```python
south.row(4).item(1)
```

## Changing the value in a cell

```python
blocks = south.column('blocks from campus').copy()
blocks[4] = 8
south = south.with_column('blocks from campus', blocks)
```
