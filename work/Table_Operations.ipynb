## Group or Pivot?

```python
skyscrapers = Table.read_table('skyscrapers.csv')
skyscrapers
```

**task**: for each city, find the tallest building per material

```python
skyscrapers.pivot('material', 'city', 'height', collect=max)
```

```python
def take_first(a):
    return a.item(0)

sky_pivot = (skyscrapers.sort('height', descending=True).pivot('material', 'city', values='name', collect=take_first))
sky_pivot
```

```python
skyscrapers.sort('height', descending=True ).select('material', 'city', 'height', 'name').group(['city', 'material'], take_first)
```

```python
skyscrapers.group?
```

```python
skyscrapers.where('city', 'Austin').sort('height', descending=True)
```

**Task:** for each city find the height difference between the talles steel building and the tallest concrete building

```python
skyscrapers.select('city', 'material', 'height').group(['city', 'material'], max)
```

```python
sky_pivot = skyscrapers.select('city', 'material', 'height').pivot('material', 'city', 'height', max)
sky_pivot
```

```python
sky_pivot = sky_pivot.with_column('difference', abs(sky_pivot.column('steel') - sky_pivot.column('concrete')))
sky_pivot
```

## Joins

```python
full = Table.read_table('nc-est2019-agesex-res.csv')
census = full.select('SEX', 'AGE', 'POPESTIMATE2019')
census
```

```python
census.group('SEX')
```

```python
sex_codes = Table().with_columns(
    'SEX CODE', make_array(0,1,2),
    'SEX descriptor', make_array('All', 'Male', 'Female')
)
sex_codes
```

```python
census.join('SEX', sex_codes, 'SEX CODE').sort('AGE')
```
