## Shortest trip between each pair of stations

```python
duration = commute.select('Start Station', 'End Station', 'Duration')
duration.show(5)
```

```python
shortest_trip = duration.group(['Start Station', 'End Station'], min)
shortest_trip
```

**Task**: find the 5 stations closest to Civic Center BART by minimum trip time

```python
cc = 'Civic Center BART (7th at Market)'
from_cc = shortest_trip.where('Start Station', cc).sort('Duration min')
from_cc.show(5)
```

# Maps

```python
# getting geographical data on the stations
stations = Table.read_table('station.csv')
stations
```

```python
stations = stations.drop(4, 6)
stations
```

```python
# stations in san fran
sf_stations = stations.where('landmark' , 'San Francisco')
sf_stations_map_data = (sf_stations
                        .select('lat', 'long', 'name',)
                        .relabeled('name', 'labels')
                       )
sf_stations_map_data.show(3)
```

```python
# map the table
Marker.map_table(sf_stations_map_data)
```

**Task** Map all stations within 4 minutes of civic center

```python
from_cc.show(3)
```

```python
close_cc = from_cc.where('Duration min', are.below(4 * 60))
close_markers = (sf_stations_map_data
                 .join('labels', close_cc, 'End Station')
                 .select('lat', 'long', 'labels')
                )
Marker.map_table(close_markers)
```

## If Statements

```python
def sign(x):
    """determines the sign (+/-) of a numeric input"""
    if x > 0:
        return 'positive'
    elif x < 0:
        return 'negative'
    elif x == 0:
        return 'neither'
```

```python
sign (-1)
```

```python
sum(np.arange(4) <= 1) * -2
```
