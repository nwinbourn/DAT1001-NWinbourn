## what is the changes of 50 heads in 100 coin flips

```python
hundred_tosses = np.random.choice(coin, 100)
num_heads = np.count_nonzero(hundred_tosses == 'heads')
num_heads
```

```python
def one_simulated_value():
    hundred_tosses = np.random.choice(coin, 100)
    num_heads = np.count_nonzero(hundred_tosses == 'heads')
    return num_heads
```

```python
one_simulated_value()
```

```python
num_reps = 10000

heads_arr = make_array()

for i in np.arange(num_reps):
    new_val = one_simulated_value()
    heads_arr = np.append(heads_arr, new_val)

heads_arr
```

```python
sim_results = Table().with_columns (
    'repitition', np.arange(1, num_reps + 1),
    'number of heads', heads_arr
)
sim_results
```

```python
sim_results.hist('number of heads')
```
